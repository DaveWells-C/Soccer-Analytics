```python
import pandas as pd

# Simple 5-player Orange County SC dataset
ocsc = pd.DataFrame({
    "Player": ["Yaniv Bazini", "Lyam MacKinnon", "Christopher Hegardt", "Ousmane Sylla", "Bryce Jamison"],
    "Minutes":[2360, 1949, 1841, 1693, 1439],
    "Goals":[10, 6, 5, 4, 1],
    "xG": [9.4, 8.9, 2.5, 3.2, 2.1]
})

display(ocsc)
```

```python
# 1. Goals minus xG (positive = clinical finishing, negative = unlucky)
ocsc["Goals_Above_xG"] = ocsc["Goals"] - ocsc["xG"]

# 2. Goals per 90: divide Goals by Minutes, then multiply by 90
ocsc["Goals_per_90"] = (ocsc["Goals"] / ocsc["Minutes"] ) * 90

# Round numbers to 2 decimal places and show the table
ocsc = ocsc.round(2)
display(ocsc)
```

```python
ocsc.to_csv("ocsc_simple_practice.csv", index=False)
print("Exported ocsc_simple_practice.csv!")
```
